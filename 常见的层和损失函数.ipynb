## 常见的函数层
1.linear(input,output): $W*x+B$<br>
2.sigmoid(input)：$\sigma(x)=\frac{1}{1+e^{-x}}$    ---输出到[0,1]<br>
3.Softmax()：$\frac{e^{x_i}}{\sum_{j=1}^{n}e^{x_j}}$<br>

## 常见损失函数
1.MSE->均方误差：回归任务<br>
$$
L_{\mathrm{MSE}}
=
\frac{1}{N}
\sum_{i=1}^{N}
(y_i-\hat y_i)^2
$$
2.MAE->平均绝对误差：回归任务<br>
$$
L_{\mathrm{MAE}}
=
\frac{1}{N}
\sum_{i=1}^{N}
|y_i-\hat y_i|
$$
3.BCE->二元交叉熵：二分类任务<br>
$$
L_{\mathrm{BCE}}
=
-\frac{1}{N}
\sum_{i=1}^{N}
\left[
y_i\log(\hat y_i)
+
(1-y_i)\log(1-\hat y_i)
\right]
$$
4.Cross Entropy->多分类交叉熵：多分类任务<br>（交叉熵损失用于衡量模型预测的概率分布与真实标签之间的差距）
$$
L_{\mathrm{CE}}
=
-\frac{1}{N}
\sum_{i=1}^{N}
\sum_{c=1}^{C}
y_{ic}\log(\hat y_{ic})
$$
