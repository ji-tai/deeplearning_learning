## 2.在np中实现多分类

````python
import numpy as np

y = np.array([1, 0, 0])
z = np.array([0.2, 0.1, -0.1])#输出的中间值z=Σsigmoid(linear())
#1.首先将输出的所有值映射到[0,1]之间，同时将使得所有的值的和变为1
y_pred = np.exp(z) / np.exp(z).sum()#将结果中的一系列值变为[0,1]之间的分布

loss = (-y * np.log(y_pred)).sum()

print(loss)
````

## 3.使用torch实现多分类

````python
import torch

y = torch.LongTensor([0])
z = torch.Tensor([[0.2, 0.1, -0.1]])
criterion = torch.nn.CrossEntropyLoss()
loss = criterion(z, y)
print(loss)
#---在torch中存在torch.nn.CrossEntropyLoss()直接集合了softmax,log,loss
#注意，y必须是LongTensor这种形式
#注意，而且z不需要再做一步激活处理，这里的z就是原始的logits，不需要sigmoid激活处理
````
